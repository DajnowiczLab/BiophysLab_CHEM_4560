# The Dawn of Quantum Theory
<span style="color: red;">**By 1900 classical physics could describe the trajectory of planets, cannonballs, and falling apples. In the early twentieth century, a series of experiments revealed that classical physics failed to describe the behavior of matter and radiation at atomic length scales.**</span><br>

**Classical picture (Newton + Maxwell)**:
- Matter is made of particles with a definite position $q$ and momentum $p$ at every instant
- Light is a continuous electromagnetic wave
- Energy can take on any value (continuous)

### 1. Black Body Radiation
A black body absorbs and emits all frequencies.

**Classical prediction (Rayleigh-Jeans)**:<br><br>
   $\rho(\nu, T) = \frac{8\pi\nu^2}{c^3}k_BT$<br>
- Works at low frequency, but $\rho \to \infty$ as $\nu \to \infty$: the **ultraviolet catastrophe**
- Total emitted energy would be infinite

**Planck (1900)**: oscillator energies are restricted to integer multiples of a quantum<br><br>
$E = nh\nu, \quad n = 0, 1, 2, ...$<br><br>
$\rho(\nu, T) = \frac{8\pi h\nu^3}{c^3}\frac{1}{e^{h\nu/k_BT} - 1}$<br>
- High-frequency oscillators cannot be excited because $h\nu \gg k_BT$
- $h = 6.626 \times 10^{-34}$ J s was a fitting parameter;

```python
import numpy as np
import plotly.graph_objects as go
from scipy.constants import h, c, k

nu = np.linspace(1e12, 1.5e15, 500)


def rayleigh_jeans(nu, T):
    return 8 * np.pi * nu**2 / c**3 * k * T


def planck(nu, T):
    return 8 * np.pi * h * nu**3 / c**3 / np.expm1(h * nu / (k * T))


fig = go.Figure()
for T, color in ((3000, 'darkorange'), (5000, 'royalblue')):
    fig.add_scatter(x=nu, y=planck(nu, T), name=f'Planck {T} K', line_color=color)
    fig.add_scatter(x=nu, y=rayleigh_jeans(nu, T), name=f'Rayleigh-Jeans {T} K', line=dict(color=color, dash='dash'))
fig.update_layout(width=750, height=450, xaxis_title='Frequency (Hz)',
                  yaxis_title='ρ(ν, T)  (J m⁻³ Hz⁻¹)', yaxis_range=[0, 2 * planck(nu, 5000).max()])
fig.show()
```

### 2. The Photoelectric Effect
Shining light on a metal ejects electrons.

**Classical prediction**: brighter light carries more energy, so electrons should be ejected faster. Any frequency should work if the light is intense enough.

**Observed**:
- No electrons below a threshold frequency $\nu_0$, no matter how bright
- Above $\nu_0$, electron kinetic energy increases linearly with frequency, not intensity
- Intensity only changes *how many* electrons are ejected

**Einstein (1905)**: light itself comes in packets (photons) of energy $h\nu$<br>
$KE_{electron} = h\nu - \Phi$<br>
where $\Phi$ is the work function of the metal.

### 3. Atomic Spectra and the Stability of Atoms

**Observed**: atoms are stable, and excited gases emit light only at discrete wavelengths. For hydrogen (Balmer, Rydberg):<br>
$\frac{1}{\lambda} = R_H\left(\frac{1}{n_1^2} - \frac{1}{n_2^2}\right)$<br>

### Takeaways
| Experiment | Classical assumption that fails | Quantum idea introduced |
|---|---|---|
| Black body radiation | Energy is continuous | $E = nh\nu$ |
| Photoelectric effect | Light is only a wave | Photons, $E = h\nu$ |
| Atomic spectra | Can have any energy | Quantized energy levels |

# Math Review
<span style="color: red;">**Quantum mechanics replaces trajectories with probability distributions, and replaces physical quantities with linear operators.**</span>

<span style="color: red;">**Is Quantum mechanics math or physics?**</span>

<span style="color: red;">*"quantum mechanics is the operating system that other physical theories run on as application software."* <br> —Scott Aaronson</span>

### Probability: Discrete Outcomes
Start with an experiment that has a finite list of possible results $x_1, x_2, ..., x_N$ (e.g. the number of diamonds in a deck of cards). Each outcome has a probability $p_i$.

$0 \le p_i \le 1$<br> <br>
$\sum_i p_i = 1$ <br> <br>


**Expectation value**: the average result over many repetitions, each outcome weighted by how often it occurs<br>
$\langle x \rangle = \sum_i x_i\,p_i$

**Variance**: the spread around the average<br>
$\sigma_x^2 = \langle (x - \langle x \rangle)^2 \rangle = \langle x^2 \rangle - \langle x \rangle^2$

### Probability: Continuous Densities
Now let the result be any real number (the position of a particle). The probability of hitting *exactly* $x = 1.000...$ is zero, so instead of a probability per outcome we use a **probability density** $p(x)$: probability per unit length.

- $p(x)\,dx$ = probability of finding $x$ in the interval $[x, x + dx]$
- Probability of finding $x$ between $a$ and $b$: $\int_a^b p(x)\,dx$
- The particle has to be *somewhere*, so the density is **normalized**:<br>
$\int_{-\infty}^{\infty} p(x)\,dx = 1$

Every discrete formula carries over with $p_i \to p(x)\,dx$ and $\sum_i \to \int dx$:

| | discrete | continuous |
|---|---|---|
| normalization | $\sum_i p_i = 1$ | $\int p(x)\,dx = 1$ |
| expectation value | $\langle f(x) \rangle = \sum_i f(x_i)\,p_i$ | $\langle f(x) \rangle = \int f(x)\,p(x)\,dx$ |
| variance | $\sigma_x^2 = \langle x^2 \rangle - \langle x \rangle^2$ | same |

**Example**: the Gaussian (normal) distribution<br>
$p(x) = \frac{1}{\sigma\sqrt{2\pi}}\,e^{-(x-\mu)^2/2\sigma^2}$<br>

```python
import numpy as np
from scipy.integrate import quad

mu, sigma = 1.0, 0.5


def gaussian(x):
    return np.exp(-(x - mu)**2 / (2 * sigma**2)) / (sigma * np.sqrt(2 * np.pi))


norm = quad(gaussian, -np.inf, np.inf)[0]
x_avg = quad(lambda x: x * gaussian(x), -np.inf, np.inf)[0]
x2_avg = quad(lambda x: x**2 * gaussian(x), -np.inf, np.inf)[0]

print(f"∫p(x)dx = {norm:.6f}")
print(f"<x>     = {x_avg:.4f}   (μ = {mu})")
print(f"σ       = {np.sqrt(x2_avg - x_avg**2):.4f}   (σ = {sigma})")
```

### Vectors and Vector Spaces
A **vector** is a list of numbers, $\mathbf{v} = (v_1, v_2, ..., v_n)$. Geometrically it is an arrow with length and direction.

A **vector space** is any set of objects you can add together and multiply by scalars and stay in the set:
- $\mathbf{u} + \mathbf{v}$ is a vector
- $c\,\mathbf{v}$ is a vector
- Arrows in 3D are a vector space. So are polynomials, and so are functions $f(x)$ (this is the one QM uses).

**Basis**: A set of linearly independent vectors $\{\mathbf{e}_1, \mathbf{e}_2, ...\}$ that spans a given vector space. In other words, such that any vector in a vector space can be written as a linear combination<br>
$\mathbf{v} = \sum_i c_i\,\mathbf{e}_i$<br>
In 3D, $\mathbf{e}_x, \mathbf{e}_y, \mathbf{e}_z$ are a basis and the $c_i$ are the components.

**Inner (dot) product**: a number that measures how much two vectors overlap<br>
$\mathbf{u}\cdot\mathbf{v} = \sum_i u_i v_i = |\mathbf{u}||\mathbf{v}|\cos\theta$
- $\mathbf{u}\cdot\mathbf{v} = 0$: **orthogonal** (perpendicular)
- $\mathbf{v}\cdot\mathbf{v} = |\mathbf{v}|^2 = 1$: **normalized**
- A basis that is both is **orthonormal**: $\mathbf{e}_i\cdot\mathbf{e}_j = \delta_{ij}$

For functions the sum becomes an integral: $\langle f | g \rangle = \int f^*(x)\,g(x)\,dx$. Normalization of $\Psi$ is just $\langle \Psi | \Psi \rangle = 1$.

### Matrices and Matrix-Vector Multiplication
A **matrix** $A$ is a grid of numbers. Multiplying a vector by a matrix produces a new vector:<br>
$(A\mathbf{v})_i = \sum_j A_{ij}\,v_j$<br>

$$
\begin{pmatrix} 2 & 0 \\ 0 & 1 \end{pmatrix}
\begin{pmatrix} 1 \\ 1 \end{pmatrix}
=
\begin{pmatrix} 2\cdot1 + 0\cdot1 \\ 0\cdot1 + 1\cdot1 \end{pmatrix}
=
\begin{pmatrix} 2 \\ 1 \end{pmatrix}
$$

A matrix is a **linear operator**: it stretches, rotates, or reflects vectors, and it respects addition and scaling<br>
$A(c_1\mathbf{u} + c_2\mathbf{v}) = c_1 A\mathbf{u} + c_2 A\mathbf{v}$


```python
def rotation(theta):
    c, s = np.cos(theta), np.sin(theta)
    return np.array([[c, -s],
                     [s,  c]])


v = np.array([1.0, 0.5])
R = rotation(np.radians(60))

print("R @ v          =", np.round(R @ v, 4))
print("|v|, |R v|     =", np.linalg.norm(v).round(4), np.linalg.norm(R @ v).round(4))
print("Rᵀ R = I       :", np.allclose(R.T @ R, np.eye(2)))
print("R(30°) R(30°) = R(60°):", np.allclose(rotation(np.radians(30)) @ rotation(np.radians(30)), R))

fig = go.Figure()
for deg in (0, 30, 60, 90, 120):
    w = rotation(np.radians(deg)) @ v
    fig.add_scatter(x=[0, w[0]], y=[0, w[1]], mode='lines+markers', name=f'R({deg}°) v',
                    marker=dict(size=[0, 10], symbol='arrow', angleref='previous'))
fig.update_layout(width=600, height=380, title='Rotating v = (1, 0.5)',
                  xaxis=dict(range=[-1.3, 1.3], constrain='domain'),
                  yaxis=dict(range=[0, 1.3], scaleanchor='x', constrain='domain'))
fig.show()
```

**Eigenvectors and eigenvalues**: special vectors that a matrix only *stretches*, without changing direction<br>
$A\mathbf{v} = \lambda\mathbf{v}$<br>
- $\lambda$ is the eigenvalue, $\mathbf{v}$ the eigenvector
- For a symmetric (Hermitian) matrix the eigenvalues are real and the eigenvectors form an orthonormal basis

Compare to the time independent Schrödinger equation, $\hat{H}\Psi = E\Psi$. It is an eigenvalue equation: $\hat{H}$ is the operator, $\Psi$ the eigenvector, and $E$ the eigenvalue. Most of computational quantum chemistry is building a matrix for $\hat{H}$ and diagonalizing it.

```python
A = np.array([[2.0, 1.0],
              [1.0, 2.0]])
v = np.array([1.0, 0.0])

print("A @ v      =", A @ v)
print("A(u+v)     =", A @ (v + np.array([0, 1])), "= Au + Av =", A @ v + A @ np.array([0, 1]))

evals, evecs = np.linalg.eigh(A)
print("\neigenvalues :", evals)
for lam, vec in zip(evals, evecs.T):
    print(f"A v = {A @ vec}   λ v = {lam * vec}")
print("orthonormal :", np.allclose(evecs.T @ evecs, np.eye(2)))
```

# The Schrödinger Equation and the Postulates of Quantum Mechanics
<span style="color: red;">**If a particle is a wave, what equation does the wave obey, and what does the wave mean?**</span><br>

### Wave-Particle Duality 
**de Broglie (1924)**: if light waves act like particles, perhaps particles act like waves<br><br>

$\lambda = \frac{h}{p} $<br><br><br>


**Double slit**:

![double slit](../assets/images/double_slit.png)


**Interference pattern are observed for particles:**<br>
A single electron somehow passes through both slits and interferes with itself.

[Electron Double Slit Experiment (Hitachi)](https://www.youtube.com/watch?v=PanqoHa_B6c)

---

<span style="color: red;">**We're going to build our understanding from basic postulates as we dive into the technical side of quantum mechanics**</span><br>

<span style="color: red;">**Postulate is statement taken to be true without proof**</span><br>



### Postulate 1: The Wavefunction
**The state of a system is completely described by a wavefunction $\Psi(x, t)$.**

Everything that can be known about the particle is contained in $\Psi$.

**Born interpretation**: $\Psi$ itself is not observable, but<br>
$|\Psi(x,t)|^2\,dx = \Psi^*\Psi\,dx$ = probability of finding the particle in $[x, x+dx]$ at time $t$ <br><br>
$\int \Psi^*\Psi\,d\tau = 1$

$\Psi$ is in general complex. The phase cannot be measured directly, but it controls how waves interfere.

### Postulate 2: Observables are Operators
**Every measurable quantity (observable) corresponds to a linear, Hermitian operator.**

An operator is an instruction that turns one function into another. You build the quantum operator by writing the classical expression in terms of $x$ and $p$ and substituting:

| Observable | Classical | Operator |
|---|---|---|
| Position | $x$ | $\hat{x} = x$ (multiply by $x$) |
| Momentum | $p$ | $\hat{p} = -i\hbar\frac{\partial}{\partial x}$ |
| Kinetic energy | $\frac{p^2}{2m}$ | $\hat{K} = \frac{\hat{p}^2}{2m} = -\frac{\hbar^2}{2m}\frac{\partial^2}{\partial x^2}$ |
| Potential energy | $V(x)$ | $\hat{V} = V(x)$ (multiply by $V$) |
| Total energy | $K + V$ | $\hat{H} = -\frac{\hbar^2}{2m}\frac{\partial^2}{\partial x^2} + V(x)$ |

- **Linear**: $\hat{O}(c_1 f + c_2 g) = c_1\hat{O}f + c_2\hat{O}g$, just like a matrix acting on vectors
- **Hermitian**: $\int f^*\hat{O}g\,dx = \int (\hat{O}f)^* g\,dx$, the function analogue of a symmetric matrix. This guarantees the properties in Postulate 3
- Operators do not always commute: $\hat{x}\hat{p} - \hat{p}\hat{x} = i\hbar$. Non-commuting observables cannot be known simultaneously (uncertainty principle)

### Postulate 3: Measurement Gives Eigenvalues
**The only possible result of measuring an observable $\hat{O}$ is one of its eigenvalues $o_n$, where $\hat{O}\phi_n = o_n\phi_n$.**

Because $\hat{O}$ is Hermitian (Postulate 2):
- All eigenvalues $o_n$ are **real**
- Eigenfunctions with different eigenvalues are **orthogonal**, and can be normalized: $\int \phi_m^*\phi_n\,dx = \delta_{mn}$
- The eigenfunctions form a **complete basis**: any allowed $\Psi$ can be written as $\Psi = \sum_n c_n\phi_n$

Compare with the math review: a symmetric matrix has real eigenvalues and orthonormal eigenvectors that span the space. Same statement, functions instead of vectors.

Immediately after a measurement returns $o_n$, the system is in the state $\phi_n$ (the wavefunction "collapses").

### Postulate 4: Expectation Values
**If the system is in state $\Psi$, the average of many measurements of $\hat{O}$ is**<br><br>
$\langle {O} \rangle = \int \Psi^*\,\hat{O}\,\Psi\,dx$

This is the expectation value from the math review, $\langle f \rangle = \int f\,p(x)\,dx$, with the operator sandwiched between $\Psi^*$ and $\Psi$ because it may contain derivatives. For position it reduces to $\langle x \rangle = \int x\,|\Psi|^2\,dx$.

Two cases:
- $\Psi$ **is an eigenfunction** $\phi_n$: every measurement gives $O_n$, so $\langle {O} \rangle$ with zero spread
- $\Psi$ **is a superposition** $\sum_n c_n\phi_n$: each measurement gives some $o_n$, with probability $|c_n|^2$, and<br>
$\langle O \rangle = \sum_n |c_n|^2\,o_n, \qquad \sum_n |c_n|^2 = 1$

The spread of measurements is $\sigma_O^2 = \langle O^2 \rangle - \langle O \rangle^2$.

### Postulate 5: Time Evolution
**The wavefunction evolves in time according to the time dependent Schrödinger equation**<br><br>
$i\hbar\frac{\partial \Psi(x,t)}{\partial t} = \hat{H}\Psi(x,t)$

(A sixth postulate, that electron wavefunctions must be antisymmetric under exchange of two electrons, is what gives the Pauli exclusion principle. It only matters once we have more than one electron, so we return to it later.)

### Putting It Together
To find the allowed energies and states of a system:
1. Write down the classical energy $K + V$ and convert to $\hat{H}$ (Postulate 2)
2. Solve $\hat{H}\psi = E\psi$ subject to the conditions on $\psi$ from Postulate 1
3. The eigenvalues are the energies you can measure (Postulate 3), and $|\psi_n|^2$ tells you where the particle is

**Atomic units**: from here on $\hbar = m_e = e = 1$. Lengths are in bohr (0.529 Å) and energies in hartree (27.2 eV), so<br>
$\hat{H} = -\frac{1}{2}\frac{d^2}{dx^2} + V(x)$

We will apply the recipe first to a particle with no potential, then to three confining potentials.

### The Free Particle
Start with the simplest case, $V(x) = 0$ everywhere:<br>
$-\frac{1}{2}\frac{d^2\psi}{dx^2} = E\psi$

Any function whose second derivative is proportional to itself works:<br>
$\psi(x) = A e^{ikx}, \qquad E = \frac{k^2}{2} = \frac{p^2}{2m}, \qquad p = \hbar k = \frac{h}{\lambda}$

- A **plane wave** with wavelength $\lambda = 2\pi/k$: de Broglie's relation falls out of the equation
- $E$ can take **any** value $\ge 0$. Nothing is quantized
- $|\psi|^2 = |A|^2$ is constant: the particle is equally likely to be anywhere. Momentum is known exactly, position not at all
- $\int |\psi|^2 dx = \infty$, so a single plane wave is not normalizable. A real free particle is a **wave packet**, a sum of plane waves with nearby $k$

**Wave packets**: adding plane waves with a spread of $k$ gives a localized $|\Psi|^2$
- Narrower in $x$ requires a wider spread in $k$ (and so in $p$): $\sigma_x\sigma_p \ge \hbar/2$, the **Heisenberg uncertainty principle**
- The packet moves at the classical velocity $p/m$ and spreads with time because each $k$ component travels at a different speed
- $|\Psi|^2$ of a Gaussian packet is a smooth bump. The oscillation is in $\Psi$ itself: a carrier wave of wavelength $2\pi/k_0$ inside the envelope $|\Psi|$

```python
import numpy as np
import plotly.graph_objects as go

x = np.linspace(-15, 60, 1200)
k = np.linspace(0, 4, 400)
k0, sigma_k = 2.0, 0.25                        
times = np.linspace(0, 20, 41)


def wave_packet(x, t):
    amp = np.exp(-(k - k0)**2 / (2 * sigma_k**2))
    psi = np.exp(1j * (x[:, None] * k - k**2 / 2 * t)) @ amp
    return psi / np.sqrt(np.trapezoid(abs(psi)**2, x))


def traces(t, **kw):
    psi = wave_packet(x, t)
    return [go.Scatter(y=psi.real, name='Re(Ψ)', **kw),
            go.Scatter(y=abs(psi), name='envelope |Ψ|', line_dash='dash', **kw)]


fig = go.Figure(data=traces(0, x=x),                      # frames only update y; x is kept from here
                frames=[go.Frame(data=traces(t), name=f'{t:.1f}') for t in times])
fig.update_layout(
    width=750, height=420, xaxis_title='x (bohr)', yaxis_title='Ψ', yaxis_range=[-0.45, 0.45],
    updatemenus=[dict(type='buttons', x=0, y=1.2, buttons=[
        dict(label='Play', method='animate', args=[None, dict(frame=dict(duration=80), fromcurrent=True)]),
        dict(label='Pause', method='animate', args=[[None], dict(mode='immediate')])])],
    sliders=[dict(currentvalue=dict(prefix='t = '), steps=[
        dict(label=f.name, method='animate', args=[[f.name], dict(mode='immediate', frame=dict(duration=0))])
        for f in fig.frames])])
fig.show()
```

### Bound States
Now add a potential that confines the particle. The particle is **bound** when $E < V$ as $x \to \pm\infty$: classically it could never escape.

Three toy potentials:

| System | $V(x)$ | $E_n$ (atomic units) |
|---|---|---|
| Particle in a box (infinite walls) | $0$ inside a box of width $L$, $\infty$ outside | $\frac{n^2\pi^2}{2L^2}$ |
| Finite square well | $0$ inside, $V_0$ outside | No closed form; finite number of states, each below the box value |
| Harmonic oscillator | $\frac{1}{2}\omega^2x^2$ | $\omega\left(n + \frac{1}{2}\right)$, $n = 0, 1, 2, ...$ |

```python

import numpy as np
import scipy.sparse as sp
from scipy.sparse.linalg import eigsh
 
 
def _laplacian_1d(n, dx):
    """Three-point Laplacian with Dirichlet boundaries, shape (n, n)."""
    return sp.diags([1.0, -2.0, 1.0], [-1, 0, 1], shape=(n, n)) / dx**2
 
 
def hamiltonian(axes, V):
    """
    Sparse Hamiltonian H = -1/2 laplacian + V on the grid spanned by `axes`.
 
    axes : list of 1D uniform grids, e.g. [x], [x, y], or [x, y, z]
    V    : array of shape (len(x), len(y), ...)  OR callable V(*mesh) -> array,
           where mesh = np.meshgrid(*axes, indexing="ij")
    """
    axes = [np.asarray(a, dtype=float) for a in axes]
    shape = tuple(a.size for a in axes)
 
    if callable(V):
        V = V(*np.meshgrid(*axes, indexing="ij"))
    V = np.asarray(V, dtype=float)
    assert V.shape == shape, f"V has shape {V.shape}, grid has shape {shape}"
 
    lap = sp.csr_matrix((np.prod(shape), np.prod(shape)))
    for d, a in enumerate(axes):
        dx = a[1] - a[0]
        term = sp.identity(1, format="csr")
        for e, n in enumerate(shape):
            block = _laplacian_1d(n, dx) if e == d else sp.identity(n)
            term = sp.kron(term, block, format="csr")
        lap = lap + term
 
    return -0.5 * lap + sp.diags(V.ravel()), shape
 
 
def bound_states(axes, V, n_states=5):
    """
    Lowest n_states eigenvalues and eigenfunctions of -1/2 laplacian + V.
 
    Returns
    -------
    E   : (n_states,) energies in hartree, ascending
    psi : (n_states, *shape) real eigenfunctions, normalised so that
          sum(|psi|^2) * prod(dx_i) = 1
    """
    H, shape = hamiltonian(axes, V)
    E, vecs = eigsh(H, k=n_states, which="SA")
    order = np.argsort(E)
    E, vecs = E[order], vecs[:, order]
 
    dV = np.prod([a[1] - a[0] for a in axes])
    vecs /= np.sqrt(np.sum(vecs**2, axis=0) * dV)
    psi = vecs.T.reshape(n_states, *shape)
    return E, psi


def probability_density(psi):
    """|psi|^2 for eigenfunctions from bound_states; integrates to 1 over the grid."""
    return np.abs(psi)**2
```

```python
from plotly.subplots import make_subplots

x = np.linspace(-8, 8, 801)
L, V0, omega = 4.0, 2.0, 1.0
potentials = {
    f'Infinite box, L = {L}':   lambda x: np.where(abs(x) < L / 2, 0.0, 1e4),
    f'Finite well, V₀ = {V0}':  lambda x: np.where(abs(x) < L / 2, 0.0, V0),
    f'Harmonic, ω = {omega}':   lambda x: 0.5 * omega**2 * x**2,
}
analytic = {0: (np.arange(1, 4)**2 * np.pi**2 / (2 * L**2)), 2: omega * (np.arange(3) + 0.5)}

fig = make_subplots(rows=1, cols=3, subplot_titles=list(potentials))
for col, (name, V) in enumerate(potentials.items()):
    E, psi = bound_states([x], V, n_states=3)
    print(f"{name:22s} E = {np.round(E, 4)}", f"  analytic {np.round(analytic[col], 4)}" if col in analytic else "")
    fig.add_scatter(x=x, y=np.minimum(V(x), 4), line_color='black', showlegend=False, row=1, col=col + 1)
    for n, (En, dn) in enumerate(zip(E, probability_density(psi))):   # draw each |ψ_n|² on its energy level
        fig.add_scatter(x=x, y=En + dn, row=1, col=col + 1)
fig.update_layout(width=950, height=400, showlegend=False, yaxis_title='Energy (hartree)', yaxis_range=[0, 4])
fig.update_xaxes(title_text='x (bohr)')
fig.show()
```

**Things to notice**
- Box: $E_n \propto n^2$, so the spacing grows.
- Finite well: same shapes as the box but every level is lower, and the tails leak into the walls (tunneling)
- Harmonic oscillator: evenly spaced levels

# The Hydrogen Atom
<span style="color: red;">**One electron and one proton: the only real atom we can solve exactly, and the source of every orbital picture in chemistry.**</span><br>

$\hat{H} = -\frac{1}{2}\nabla^2 - \frac{Z}{r}$

**Solve using Spherical coordinates** <br> <br>
$\nabla^2  = \frac{1}{r^2}\frac{\partial}{\partial r}\left(r^2 \frac{\partial }{\partial r}\right) + \frac{1}{r^2 \sin\theta}\frac{\partial}{\partial \theta}\left(\sin\theta \frac{\partial }{\partial \theta}\right) + \frac{1}{r^2 \sin^2\theta}\frac{\partial^2 }{\partial \phi^2}$




### The Angular Part: Spherical Harmonics
$Y_{lm}(\theta, \phi)$ are the eigenfunctions of angular momentum

- $l = 0, 1, 2, 3, ...$ is the **angular momentum quantum number**.
- $m = -l, ..., 0, ..., +l$ is the **magnetic quantum number**,


| $l$ | letter | $2l+1$ | shape |
|---|---|---|---|
| 0 | s | 1 | sphere, no angular nodes |
| 1 | p | 3 | two lobes, one nodal plane |
| 2 | d | 5 | four lobes (or two lobes and a ring), two nodal surfaces |
| 3 | f | 7 | three nodal surfaces |


### The Radial Part

| orbital | $n$ | $l$ | $R_{nl}(r)$ | radial nodes |
|---|---|---|---|---|
| 1s | 1 | 0 | $2\,e^{-r}$ | 0 |
| 2s | 2 | 0 | $\frac{1}{2\sqrt{2}}(2 - r)\,e^{-r/2}$ | 1 |
| 2p | 2 | 1 | $\frac{1}{2\sqrt{6}}\,r\,e^{-r/2}$ | 0 |
| 3s | 3 | 0 | $\frac{2}{81\sqrt{3}}(27 - 18r + 2r^2)\,e^{-r/3}$ | 2 |
| 3p | 3 | 1 | $\frac{4}{81\sqrt{6}}(6 - r)\,r\,e^{-r/3}$ | 1 |
| 3d | 3 | 2 | $\frac{4}{81\sqrt{30}}\,r^2\,e^{-r/3}$ | 0 |

**Radial distribution function**. The probability of finding the electron between $r$ and $r + dr$, in any direction, is<br>
$P(r)\,dr = r^2\,|R_{nl}(r)|^2\,dr$<br>
The $r^2$ comes from the volume element: there is more space in a shell at large $r$. For 1s, $|\psi|^2$ is largest at the nucleus but $P(r)$ peaks at $r = 1$ bohr, the Bohr radius.

### Orbitals in 3D
Put the pieces together: $\psi_{nlm} = R_{nl}(r)\,Y_{lm}(\theta, \phi)$, and $|\psi|^2$ is the probability density of the electron.

The surfaces below are drawn where $\psi = \pm c$, with the two signs in different colours. The sign of $\psi$ is not observable on its own (only $|\psi|^2$ is), but it decides how orbitals on neighbouring atoms add or cancel.

```python
import numpy as np
import plotly.graph_objects as go
from skimage import measure
from plotly.subplots import make_subplots
from scipy.special import genlaguerre, factorial


def R(n, l, r):
    # hydrogen radial function, atomic units
    rho = 2 * r / n
    norm = np.sqrt((2 / n)**3 * factorial(n - l - 1) / (2 * n * factorial(n + l)))
    return norm * np.exp(-rho / 2) * rho**l * genlaguerre(n - l - 1, 2 * l + 1)(rho)


r = np.linspace(0, 30, 1000)

angular = {   # real spherical harmonics as functions of x, y, z
    's':    lambda x, y, z, r: np.full_like(r, 0.5 / np.sqrt(np.pi)),
    'p_z':  lambda x, y, z, r: np.sqrt(3 / (4 * np.pi)) * z / r,
    'd_z2': lambda x, y, z, r: np.sqrt(5 / (16 * np.pi)) * (3 * z**2 - r**2) / r**2,
    'd_xy': lambda x, y, z, r: np.sqrt(15 / (4 * np.pi)) * x * y / r**2,
}
orbitals = {'1s': (1, 0, 's'), '2p_z': (2, 1, 'p_z'), '3d_z2': (3, 2, 'd_z2'), '3d_xy': (3, 2, 'd_xy')}

fig = make_subplots(rows=1, cols=4, specs=[[{'type': 'scene'}] * 4], subplot_titles=list(orbitals))
for col, (name, (n, l, ang)) in enumerate(orbitals.items(), 1):
    L = 3 * n**2 + 2                                           # box half-width, bohr
    ax = np.linspace(-L, L, 70)
    x, y, z = np.meshgrid(ax, ax, ax, indexing='ij')
    rr = np.sqrt(x**2 + y**2 + z**2) + 1e-9
    psi = R(n, l, rr) * angular[ang](x, y, z, rr)
    level = 0.3 * abs(psi).max()
    for sign, color in ((1, 'crimson'), (-1, 'royalblue')):   # one surface per sign of ψ
        if psi.min() < sign * level < psi.max():
            verts, tris, *_ = measure.marching_cubes(psi, level=sign * level)
            xyz = -L + verts * (ax[1] - ax[0])
            fig.add_mesh3d(x=xyz[:, 0], y=xyz[:, 1], z=xyz[:, 2], i=tris[:, 0], j=tris[:, 1], k=tris[:, 2],
                           color=color, opacity=0.85, row=1, col=col)
    fig.update_scenes(dict(aspectmode='cube', xaxis_visible=False, yaxis_visible=False, zaxis_visible=False),
                      row=1, col=col)
fig.update_layout(width=1000, height=320, margin=dict(l=0, r=0, t=30, b=0))
fig.show()
```

### Summary: Three Quantum Numbers
| | values | sets |
|---|---|---|
| $n$ | $1, 2, 3, ...$ | energy $E_n = -1/2n^2$, size $\sim n^2$, total nodes $n - 1$ |
| $l$ | $0, ..., n-1$ | shape (s, p, d, f) |
| $m$ | $-l, ..., +l$ | orientation|

A fourth, spin $m_s = \pm\frac{1}{2}$, is not in the Schrödinger equation at all. It comes from relativity (Dirac), and with the Pauli principle it is what fills the periodic table two electrons per orbital.

Every multi-electron atom and every molecule is described, to a first approximation, by combining these hydrogen-like orbitals.

# Molecules: $H_2^+$ and $H_2$
<span style="color: red;">**Why do two atoms stick together? What is a chemical bond in the language of wavefunctions?**</span><br>

**Starting point**: the hydrogen atom orbitals from the previous section. A molecular wavefunction is built by combining atomic orbitals on different nuclei, and the sign of $\psi$ on each atom decides whether they reinforce or cancel.

### The Born-Oppenheimer Approximation
A molecule has electrons and nuclei, and $\hat{H}$ contains the kinetic energy of both plus every pairwise Coulomb term. Nuclei are at least 1836 times heavier than electrons, so they move far more slowly.

**Approximation**: freeze the nuclei at positions $\mathbf{R}$, solve for the electrons, then move the nuclei.<br>
$\hat{H}_{el}\,\psi_{el}(\mathbf{r}; \mathbf{R}) = E_{el}(\mathbf{R})\,\psi_{el}(\mathbf{r}; \mathbf{R})$

- The electronic energy plus nuclear repulsion, $E(\mathbf{R}) = E_{el}(\mathbf{R}) + V_{NN}(\mathbf{R})$, is the **potential energy surface**
- This surface is the $V(r)$ that the force fields in Lecture 3 are approximating with springs and Lennard-Jones terms
- Bond lengths are its minima, vibrational frequencies its curvature, and reaction barriers its saddle points

### $H_2^+$: One Electron, Two Nuclei
The simplest molecule. Nuclei A and B separated by $R$, one electron:<br>
$\hat{H}_{el} = -\frac{1}{2}\nabla^2 - \frac{1}{r_A} - \frac{1}{r_B}, \qquad E(R) = E_{el}(R) + \frac{1}{R}$

**Linear Combination of Atomic Orbitals (LCAO)**. Near nucleus A the electron should look like a hydrogen 1s orbital on A, and likewise near B. So guess<br>
$\psi = c_A\,1s_A + c_B\,1s_B$

By symmetry $|c_A| = |c_B|$, leaving two possibilities:<br>
$\psi_\pm = \frac{1s_A \pm 1s_B}{\sqrt{2(1 \pm S)}}, \qquad S = \int 1s_A\,1s_B\,d\tau$

- $S$ is the **overlap integral**, 1 at $R = 0$ and 0 at $R \to \infty$
- $\psi_+$ (**bonding**, $\sigma_g$): the orbitals add. Electron density builds up between the nuclei, where it is attracted to both
- $\psi_-$ (**antibonding**, $\sigma_u^*$): the orbitals cancel on the midplane. A node between the nuclei pushes density outward

```python
import numpy as np
import plotly.graph_objects as go
from plotly.subplots import make_subplots

E_1s = -0.5


def lcao_h2plus(R):
    """Bonding and antibonding energies of H2+ in the 1s LCAO approximation (hartree)."""
    S = np.exp(-R) * (1 + R + R**2 / 3)
    alpha = E_1s - (1 / R) * (1.0 - np.exp(-2 * R) * (1 + R))
    beta = (S * E_1s) - np.exp(-R) * (1 + R) 

    nuclear_repulsion = 1 / R
    ene_plus = (alpha + beta) / (1 + S) + nuclear_repulsion
    ene_minus = (alpha - beta) / (1 - S) + nuclear_repulsion
    return ene_plus, ene_minus, S


R = np.linspace(0.8, 8, 300)
E_plus, E_minus, S = lcao_h2plus(R)
i = E_plus.argmin()
print(f"LCAO: R_e = {R[i]:.2f} bohr, D_e = {(E_1s - E_plus[i]) * 27.211:.2f} eV   (exact: 2.00 bohr, 2.79 eV)")

# orbitals along the bond axis at R = 2
z = np.linspace(-5, 5, 400)
one_s = lambda z0: np.exp(-abs(z - z0)) / np.sqrt(np.pi)
S2 = lcao_h2plus(2.0)[2]
psi_plus = (one_s(-1) + one_s(1)) / np.sqrt(2 * (1 + S2))
psi_minus = (one_s(-1) - one_s(1)) / np.sqrt(2 * (1 - S2))

fig = make_subplots(rows=1, cols=2)
fig.add_scatter(x=R, y=E_plus, name='σ_g bonding  E₊', row=1, col=1)
fig.add_scatter(x=R, y=E_minus, name='σ_u* antibonding  E₋', row=1, col=1)
fig.add_hline(y=E_1s, line_dash='dot', annotation_text='H + H⁺', row=1, col=1)
fig.add_scatter(x=z, y=psi_plus**2, name='|ψ₊|²', line_dash='dash', row=1, col=2)
fig.add_scatter(x=z, y=psi_minus**2, name='|ψ₋|²', line_dash='dash', row=1, col=2)
fig.update_xaxes(title_text='R (bohr)', row=1, col=1)
fig.update_xaxes(title_text='z (bohr)', row=1, col=2)
fig.update_yaxes(title_text='E (hartree)', range=[-0.65, -0.2], row=1, col=1)
fig.update_layout(width=950, height=400)
fig.show()
```

### $H_2$: Two Electrons
Add a second electron. The MO picture: put both electrons in the bonding orbital, $\sigma_g^2$.<br>
$\Psi(1, 2) = \psi_+(\mathbf{r}_1)\,\psi_+(\mathbf{r}_2)$

**Spin and the Pauli principle**. Electrons have spin, $\alpha$ (up) or $\beta$ (down), and the sixth postulate says the total wavefunction must change sign when two electrons are swapped. For two electrons in the same spatial orbital, the spin part must be antisymmetric:<br>
$\Psi(1, 2) = \psi_+(1)\psi_+(2)\cdot\frac{1}{\sqrt{2}}\left[\alpha(1)\beta(2) - \beta(1)\alpha(2)\right]$
- Swap 1 and 2: the spatial part is unchanged, the spin part flips sign. Antisymmetric, as required
- Two electrons can share an orbital only with opposite spins. A third electron would have to go in $\sigma_u^*$. This is the exclusion principle, and it is why the periodic table has shells
- In general the antisymmetric product is written as a **Slater determinant**

**Electron-electron repulsion**. $\hat{H}$ now contains $+\frac{1}{r_{12}}$, which couples the two electrons. The exact two-electron problem has no closed form, even with the nuclei frozen.

# Excitation: Absorbance, Emission, and the Transition Dipole
<span style="color: red;">**Why does a molecule absorb light at one wavelength and not another, and why are some transitions bright and others dark?**</span><br>

Everything so far was stationary states. Spectroscopy is about **transitions** between them, driven by light.

**Energy conservation** (Bohr frequency condition): a photon is absorbed or emitted only if<br>
$h\nu = E_f - E_i$

This tells you *where* the lines are. It does not tell you *how strong* they are.

**Light as a perturbation**. The electric field of light, $\mathbf{E}(t) = \mathbf{E}_0\cos(2\pi\nu t)$, interacts with the molecule's dipole moment $\hat{\boldsymbol{\mu}} = \sum_i q_i\,\hat{\mathbf{r}}_i$:<br>
$\hat{H}(t) = \hat{H}_0 - \hat{\boldsymbol{\mu}}\cdot\mathbf{E}(t)$

Solving the time dependent Schrödinger equation (Postulate 5)<br>

**Transition dipole moment**:<br>
$\boldsymbol{\mu}_{fi} = \int \psi_f^*\,\hat{\boldsymbol{\mu}}\,\psi_i\,d\tau$

- It looks like an expectation value (Postulate 4) but with *two different* states. It measures how much the oscillating field can mix $\psi_i$ with $\psi_f$
- It is a vector. A transition is driven only by the component of $\mathbf{E}$ along $\boldsymbol{\mu}_{fi}$
- If $\boldsymbol{\mu}_{fi} = 0$ the transition is **forbidden**

so singlet $\to$ triplet transitions are forbidden. (They happen anyway, weakly, through spin-orbit coupling. That is phosphorescence.)

"Forbidden" means weak, not impossible. Real molecules are not perfectly symmetric, and vibrations break the symmetry further.


### Absorbance
In the lab you measure transmitted intensity through a sample of concentration $c$ and path length $l$:<br>
$A = -\log_{10}\frac{I}{I_0} = \varepsilon(\lambda)\,c\,l$ (Beer-Lambert)

The molar absorptivity $\varepsilon$<br>
$\int \varepsilon(\nu)\,d\nu \propto \nu_{fi}\,|\boldsymbol{\mu}_{fi}|^2$

often quoted as the dimensionless **oscillator strength** $f \approx 4.3 \times 10^{-9}\int\varepsilon\,d\bar\nu$.

- A **chromophore** is a group whose $E_f - E_i$ falls in the UV-visible range: conjugated $\pi$ systems, where the particle in a box is a surprisingly decent model. Longer conjugation means a bigger box, smaller gap, redder absorption
- Proteins absorb at 280 nm (Trp, Tyr $\pi \to \pi^*$). Nucleic acids absorb at 260 nm. This is where $A_{280}$ and $A_{260}/A_{280}$ come from
- Transition dipole directions are fixed in the molecular frame, so oriented samples absorb polarized light anisotropically (linear dichroism, and the CD signal of protein secondary structure)

### Emission
![emission](../assets/images/emission.png)<br>
[From chemlibre text](https://chem.libretexts.org/Bookshelves/Physical_and_Theoretical_Chemistry_Textbook_Maps/Physical_Chemistry_(LibreTexts)/15%3A_Lasers_Laser_Spectroscopy_and_Photochemistry/15.02%3A_The_Dynamics_of_Transitions_can_be_Modeled_by_Rate_Equations)

**Fluorescence** is emission from the lowest excited singlet state $S_1$ to $S_0$:
1. Absorb a photon: $S_0 \to S_1$ (or higher), with the nuclei frozen (Franck-Condon, the Born-Oppenheimer idea again). The molecule lands in an excited *vibrational* level of $S_1$
2. Vibrational relaxation and internal conversion to the bottom of $S_1$, dumping energy as heat
3. Emit from the bottom of $S_1$

Because step 2 throws energy away, emission is always at lower energy than absorption: the **Stokes shift**. The emission spectrum is roughly a mirror image of the absorption spectrum.

### Summary
- Lines appear where $h\nu = E_f - E_i$
- Their strength is $|\boldsymbol{\mu}_{fi}|^2$, and parity decides whether it is zero
- Absorbance measures $|\boldsymbol{\mu}_{fi}|^2$ through $\varepsilon$; emission measures it through $\tau_{rad}$ (decay rate)
- Everything between absorption and emission (Stokes shift, quantum yield, lifetime) is the dynamics of excited state
